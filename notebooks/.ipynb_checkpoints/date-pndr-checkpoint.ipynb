```python
# Verifică dacă ai scikit-learn instalat (dacă dă eroare, rulează !pip install scikit-learn într-o celulă)
from sklearn.cluster import KMeans
import pandas as pd

# 1. Filtrăm doar rândurile care au coordonate valide
df_valid = df.dropna(subset=['latitude', 'longitude']).copy()

# 2. Aplicăm algoritmul K-Means pentru a crea 42 de "zone/județe" geografice
print("Se grupează localitățile în 42 de regiuni geografice...")
kmeans = KMeans(n_clusters=42, random_state=42, n_init=10)
df_valid['cluster_judet'] = kmeans.fit_predict(df_valid[['latitude', 'longitude']])

# 3. Adăugăm rezultatul înapoi în tabelul tău principal
df.loc[df_valid.index, 'cluster_judet'] = df_valid['cluster_judet']

# 4. Vedem câte localități a grupat în fiecare dintre cele 42 zone
print(df['cluster_judet'].value_counts())
```

```python
import matplotlib.pyplot as plt

# Setează dimensiunea graficului
plt.figure(figsize=(14, 6))

# Creează un bar chart ordonat după ID-ul clusterului
df['cluster_judet'].value_counts().sort_index().plot(kind='bar', color='teal')

plt.title('Numărul de localități din fișier grupate pe regiuni geografice (Clustere)', fontsize=14)
plt.xlabel('ID Regiune (Cluster)', fontsize=12)
plt.ylabel('Număr de localități', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Schimbă numărul 0 cu orice altă cifră de la 0 la 41 ca să vezi ce județ/zonă este
cluster_id = 0
localitati_din_cluster = df[df['cluster_judet'] == cluster_id]['localitate'].head(15)

print(f"Localități din clusterul {cluster_id}:")
print(localitati_din_cluster)# Afișează o localitate din fiecare dintre cele 42 clustere ca să le recunoaștem imediat
for cid in range(42):
    exemplu = df[df['cluster_judet'] == cid]['localitate'].values[0]
    print(f"Clusterul {cid} conține localități precum: {exemplu}")
```

```python
import pandas as pd

# Căutăm în fiecare cluster localitățile care au "municipiu" în denumire
for cid in range(42):
    subset = df[df['cluster_judet'] == cid]
    
    # Filtrăm rândurile care conțin "municipiu" (ignorând majuscule/minuscule)
    municipii = subset[subset['localitate'].astype(str).str.contains('municipiu', case=False, na=False)]['localitate'].unique()
    
    if len(municipii) > 0:
        print(f"Cluster {cid} ➔ {list(municipii)}")
    else:
        # Dacă dintr-un motiv oarecare nu găsește cuvântul exact, afișăm primele câteva localități ca fallback
         primele = subset['localitate'].head(3).values
         print(f"Cluster {cid} (fără 'municipiu' în text) ➔ Exemplu: {list(primele)}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Grupăm datele pe județe și calculăm totalurile
df_grafic = df.groupby('judet').agg(
    numar_proiecte=('judet', 'count'),
    total_contractat=('valoare_contractata', 'sum'),
    total_platit=('valoarea_platita', 'sum')
).sort_values(by='total_contractat', ascending=False) # Ordonăm descrescător după sumele contractate

# 2. Setăm dimensiunea generală a planșei (2 grafice, unul sub altul)
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 12), sharex=True)

# --- GRAFICUL 1: Numărul de proiecte ---
ax1.bar(df_grafic.index, df_grafic['numar_proiecte'], color='teal', alpha=0.8)
ax1.set_title('Numărul de proiecte pe fiecare județ', fontsize=14, fontweight='bold')
ax1.set_ylabel('Număr proiecte', fontsize=12)
ax1.grid(axis='y', linestyle='--', alpha=0.5)

# --- GRAFICUL 2: Valoare contractată vs. plătită ---
x = np.arange(len(df_grafic.index)) # Pozițiile pe axa X
latime_bara = 0.4 # Cât de groasă să fie o bară

# Desenăm două rânduri de bare, ușor decalate stânga-dreapta
ax2.bar(x - latime_bara/2, df_grafic['total_contractat'], width=latime_bara, label='Valoare Contractată', color='royalblue')
ax2.bar(x + latime_bara/2, df_grafic['total_platit'], width=latime_bara, label='Valoare Plătită', color='darkorange')

ax2.set_title('Valoare Contractată vs. Valoare Plătită', fontsize=14, fontweight='bold')
ax2.set_ylabel('Suma (Valoare)', fontsize=12)
ax2.set_xticks(x)
ax2.set_xticklabels(df_grafic.index, rotation=90) # Rotim numele județelor ca să încapă
ax2.legend()
ax2.grid(axis='y', linestyle='--', alpha=0.5)

# Ajustăm spațiul și afișăm
plt.tight_layout()
plt.show()

# (Opțional) Afișăm și tabelul cu top 10 ca să vezi cifrele exacte
display(df_grafic.head(10))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Grupăm datele pe județe, calculând numărul de proiecte și suma valorilor
df_judete = df.groupby('judet').agg(
    numar_proiecte=('judet', 'size'),
    total_contractat=('valoare_contractata', 'sum'),
    total_platit=('valoarea_platita', 'sum')
).reset_index()

# 2. Sortăm tabelul (de exemplu, descrescător după valoarea contractată)
# pentru a avea un grafic ordonat și ușor de citit
df_judete = df_judete.sort_values(by='total_contractat', ascending=False)

# Afișăm primele rânduri din tabelul centralizat pentru a verifica cifrele
display(df_judete.head())

# ==========================================
# 3. Generarea Graficelor
# ==========================================

# Creăm o figură mare cu 2 grafice așezate pe verticală (nrows=2), împărțind aceeași axă X (sharex=True)
fig, (ax1, ax2) = plt.subplots(nrows=2, ncols=1, figsize=(18, 12), sharex=True)

# Pozițiile pe axa X pentru cele ~42 de județe
x = np.arange(len(df_judete['judet']))
width = 0.4  # Lățimea barelor pentru graficul cu valori duble

# --- Graficul 1 (Sus): Numărul de proiecte ---
ax1.bar(x, df_judete['numar_proiecte'], color='teal')
ax1.set_title('Numărul de Proiecte pe Județe', fontsize=16, pad=15)
ax1.set_ylabel('Număr Proiecte', fontsize=12)
ax1.grid(axis='y', linestyle='--', alpha=0.6)

# --- Graficul 2 (Jos): Valoare Contractată vs. Valoare Plătită ---
# Decalăm barele ușor spre stânga și spre dreapta pentru a le pune una lângă alta
ax2.bar(x - width/2, df_judete['total_contractat'], width, label='Valoare Contractată', color='royalblue')
ax2.bar(x + width/2, df_judete['total_platit'], width, label='Valoare Plătită', color='darkorange')

ax2.set_title('Valoarea Contractată vs. Valoarea Plătită', fontsize=16, pad=15)
ax2.set_ylabel('Suma Totală', fontsize=12)
ax2.legend(fontsize=12)
ax2.grid(axis='y', linestyle='--', alpha=0.6)

# Setăm etichetele cu numele județelor pe axa X (rotite la 90 de grade ca să încapă toate)
ax2.set_xticks(x)
ax2.set_xticklabels(df_judete['judet'], rotation=90, fontsize=11)

# Ajustăm marginile ca să nu se taie etichetele și afișăm graficul
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Ne asigurăm că valorile sunt numerice (opțional, dar previne erorile dacă sumele sunt citite ca text)
df['valoare_contractata'] = pd.to_numeric(df['valoare_contractata'], errors='coerce').fillna(0)
df['valoare_platita'] = pd.to_numeric(df['valoare_platita'], errors='coerce').fillna(0)

# 2. Grupăm datele pe județe (numărăm rândurile pentru numărul de proiecte și însumăm banii)
df_judete = df.groupby('judet').agg(
    numar_proiecte=('judet', 'size'),
    total_contractat=('valoare_contractata', 'sum'),
    total_platit=('valoare_platita', 'sum')
).reset_index()

# Sortăm descrescător ca să arate bine vizual (de la județul cu cei mai mulți bani la cel cu cei mai puțini)
df_judete = df_judete.sort_values(by='total_contractat', ascending=False)

# 3. Desenăm graficul
fig, ax1 = plt.subplots(figsize=(18, 8)) # O lățime mai mare ca să încapă toate cele 42 de județe

# Setăm lățimea barelor și poziția lor pe axa X
x = np.arange(len(df_judete['judet']))
width = 0.35

# Desenăm barele pentru valorile financiare (pe axa din Stânga)
bara_contractat = ax1.bar(x - width/2, df_judete['total_contractat'], width, label='Valoare Contractată', color='#1f77b4')
bara_platit = ax1.bar(x + width/2, df_judete['total_platit'], width, label='Valoare Plătită', color='#2ca02c')

ax1.set_xlabel('Județ', fontsize=12)
ax1.set_ylabel('Sume (RON)', fontsize=12)
ax1.set_title('Proiecte pe Județe: Valori financiare vs. Număr de proiecte', fontsize=16)
ax1.set_xticks(x)
ax1.set_xticklabels(df_judete['judet'], rotation=45, ha='right')
ax1.legend(loc='upper left')

# 4. Adăugăm a doua axă Y (pe Dreapta) pentru numărul de proiecte
ax2 = ax1.twinx()
linie_proiecte = ax2.plot(x, df_judete['numar_proiecte'], color='red', marker='o', linewidth=2, label='Număr Proiecte')

ax2.set_ylabel('Număr de Proiecte', fontsize=12, color='red')
ax2.tick_params(axis='y', labelcolor='red')
ax2.legend(loc='upper right')

# Adăugăm un grid ușor de citit
ax1.grid(axis='y', linestyle='--', alpha=0.4)

# Afișăm graficul curat
plt.tight_layout()
plt.show()
```

```python
import folium
from folium.plugins import HeatMap
import pandas as pd

# 1. Curățăm datele (eliminăm rândurile fără coordonate)
df_harta = df.dropna(subset=['latitude', 'longitude']).copy()
df_harta['valoare_contractata'] = pd.to_numeric(df_harta['valoare_contractata'], errors='coerce').fillna(0)

# 2. Folium HeatMap are nevoie de date sub formă de listă: [lat, lon, valoare/greutate]
heat_data = df_harta[['latitude', 'longitude', 'valoare_contractata']].values.tolist()

# 3. Inițializăm harta centrată pe România (CartoDB positron este un fundal gri deschis curat)
harta_ro = folium.Map(location=[45.9, 25.0], zoom_start=6, tiles='CartoDB positron')

# 4. Adăugăm stratul de heatmap
# radius controlează mărimea petei, blur cât de difuză este
HeatMap(heat_data, radius=15, blur=10, max_zoom=1).add_to(harta_ro)

# 5. Afișăm harta direct în celulă
harta_ro
```

```python
import folium
from folium.plugins import HeatMap
import numpy as np
import pandas as pd

# 1. Curățăm datele și ne asigurăm că sunt numere
df_harta = df.dropna(subset=['latitude', 'longitude']).copy()
df_harta['valoare_contractata'] = pd.to_numeric(df_harta['valoare_contractata'], errors='coerce').fillna(0)

# Păstrăm doar proiectele care au efectiv o valoare peste 0
df_harta = df_harta[df_harta['valoare_contractata'] > 0]

# 2. TRUCUL MAGIC: Aplicăm scară logaritmică pentru a calma sumele de milioane 
# și normalizăm totul între 0 și 1 (formatul perfect pentru Folium Heatmap)
df_harta['greutate_log'] = np.log1p(df_harta['valoare_contractata']) # logaritm
max_val = df_harta['greutate_log'].max()
df_harta['greutate_normalizata'] = df_harta['greutate_log'] / max_val

# 3. Pregătim datele [lat, lon, greutate_noua]
heat_data = df_harta[['latitude', 'longitude', 'greutate_normalizata']].values.tolist()

# 4. Desenăm harta
harta_ro = folium.Map(location=[45.9, 25.0], zoom_start=6, tiles='CartoDB positron')

# Am micșorat radius și blur ca să vezi clar diferențele, nu doar o pată mare
HeatMap(heat_data, radius=8, blur=6, max_val=1.0).add_to(harta_ro)

# Afișăm harta
harta_ro
```

```python
import folium
import pandas as pd

# Coordonatele reședințelor de județ pentru a plasa bulele exact în centrul fiecărui județ
judete_resedinte = {
    'Alba': (46.0683, 23.5716), 'Arad': (46.1866, 21.3123), 'Argeș': (44.8565, 24.8692),
    'Bacău': (46.5667, 26.9167), 'Bihor': (47.0465, 21.9189), 'Bistrița-Năsăud': (47.1329, 24.4912),
    'Botoșani': (47.7486, 26.6657), 'Brașov': (45.6579, 25.6012), 'Brăila': (45.2692, 27.9575),
    'Buzău': (45.1500, 26.8167), 'Caraș-Severin': (45.3000, 21.8833), 'Călărași': (44.2000, 27.3333),
    'Cluj': (46.7712, 23.6236), 'Constanța': (44.1792, 28.6499), 'Covasna': (45.8500, 25.7833),
    'Dâmbovița': (44.9292, 25.4567), 'Dolj': (44.3302, 23.7949), 'Galați': (45.4353, 28.0077),
    'Giurgiu': (43.8903, 25.9678), 'Gorj': (45.0333, 23.2833), 'Harghita': (46.3636, 25.7997),
    'Hunedoara': (45.8833, 22.9000), 'Ialomița': (44.5667, 27.3667), 'Iași': (47.1585, 27.6014),
    'Ilfov': (44.4323, 26.1063), 'Maramureș': (47.6592, 23.5650), 'Mehedinți': (44.6333, 22.6500),
    'Mureș': (46.5456, 24.5624), 'Neamț': (46.9281, 26.3714), 'Olt': (44.4333, 24.3667),
    'Prahova': (44.9404, 26.0264), 'Satu Mare': (47.7924, 22.8836), 'Sălaj': (47.1833, 23.0500),
    'Sibiu': (45.7983, 24.1255), 'Suceava': (47.6514, 26.2556), 'Teleorman': (43.9667, 25.3333),
    'Timiș': (45.7537, 21.2257), 'Tulcea': (45.1781, 28.7964), 'Vaslui': (46.6407, 27.7276),
    'Vâlcea': (45.1000, 24.3667), 'Vrancea': (45.7000, 27.1833), 'București': (44.4268, 26.1025)
}

# 1. Ne asigurăm că banii sunt numere și adunăm totalul contractat pe fiecare județ
df['valoare_contractata'] = pd.to_numeric(df['valoare_contractata'], errors='coerce').fillna(0)
df_sume_judete = df.groupby('judet')['valoare_contractata'].sum().reset_index()

# 2. Inițializăm harta (folosim OpenStreetMap care e mereu gratuit și stabil)
harta_bule = folium.Map(location=[45.9, 25.0], zoom_start=6.5, tiles='OpenStreetMap')

# Găsim valoarea maximă din țară pentru a calcula proporțiile cercurilor
valoare_max = df_sume_judete['valoare_contractata'].max()

# 3. Desenăm cercurile
for _, row in df_sume_judete.iterrows():
    judet = row['judet']
    valoare = row['valoare_contractata']
    
    if judet in judete_resedinte and valoare > 0:
        lat, lon = judete_resedinte[judet]
        
        # Raza cercului variază între 5 și 35 în funcție de suma contractată
        raza_cerc = 5 + 30 * (valoare / valoare_max)
        
        # Adăugăm cercul pe hartă
        folium.CircleMarker(
            location=[lat, lon],
            radius=raza_cerc,
            popup=folium.Popup(f"<b>{judet}</b><br>Valoare: {valoare:,.0f} RON", max_width=250),
            tooltip=judet,  # Afișează numele județului când pui mouse-ul pe cerc
            color='crimson',
            fill=True,
            fill_color='crimson',
            fill_opacity=0.6
        ).add_to(harta_bule)

# 4. Afișăm harta
harta_bule
```
